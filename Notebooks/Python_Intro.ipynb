## Create an environment for your work

Environments are a way to manage dependencies for different projects. They allow you to create isolated spaces where you can install specific versions of packages without affecting other projects. This can help protect your base Python installation, and ensure that your code runs consistently across different machines and setups which fosters collaboration and reproducibility.

To create a new environment, you can use the following command in the terminal:
```bash
conda create --name myenv python=3.12
```

Ensure you activate your environment before installing any packages or running your code. You can activate the environment with:
```bash
conda activate myenv
```

You should see the name of your environment in parentheses at the beginning of the terminal prompt, indicating that it is active.

In VS Code, you can also select the Python interpreter associated with your environment by clicking on the Python version in the top right corner of the window and selecting your environment from the list. VS Code will typically save this setting for the current workspace, so you won't have to select it again unless you change projects.
