### Variables
A variable is a name that refers to a value stored in memory. Variables allow you to store and manipulate data by giving it a meaningful label, which can be used throughout your code. Variables are assigned values using the ```=``` operator. There are naming rules for variables:
- start with a letter (e.g. ```a-z```, ```A-Z```) or an underscore (```_```), followed by letters, numbers or underscores
- cannot contain spaces or special characters (e.g. @, £, %)
- cannot be a reserved keyword (e.g. ```if```, ```for```, ```True``` etc)

````python
x = 10       # x is an integer
print(type(x))

y = "Hello"  # y is a string
print(type(y))

z = 3.14     # z is a float
print(type(z))
````

### Comments
Comments are pieces of text within the code that are ignored by the interpreter. They are used to explain or annotate the code to make it more readable and understandable for developers, but they don't affect how the program runs. Comments are defined using the ```#``` symbol and everything following the ```#``` on that line is ignored by Python. See above.

### Strings
