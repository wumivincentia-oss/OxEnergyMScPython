### Indexing
Indexing refers to accessing individual elements within a sequence (like strings, lists, or tuples) by their position, or index. Python uses zero-based indexing, which means that the first element in a sequence is at index 0, the second element is at index 1, and so on. Indexing is typically implemented with square brackets ```[0]```.

You can index from the end of a sequence counting backwards by using negative indexes e.g. ```[-2]``` is the second element from the end.


````python
message = "Hello Energy"

print(message[0])  # First character
print(message[1])  # Second character
print(message[-1]) # Last character
````

### Slicing
Slicing is an extension of indexing that allows you to access a range of elements in a sequence.
The syntax for slicing is sequence ```[start:end]```, where ```start``` is inclusive and ```end``` is exclusive.

````python
print(message[0:5])
````

### Dot notation
Dot notation in Python is a syntax used to access attributes, methods, or properties of objects, classes, and modules. It allows you to "drill down" into an object to retrieve data or invoke behaviors associated with it.

````python
print(message.upper())  # access the string's 'upper' method
````

## Loops
Loops are used to execute a block of code repeatedly until a condition is met. There are two types:
- For loops: used to iterate over a sequence of objects
- While loops: used to execute a block of code while a certain condition is true.

### For loops
Classed as an iteration statement, it allows for part of the code to be repeatedly executed.

````python
countries = ["Namibia", "Portugal", "Egypt", "Haiti", "Thailand", "Bolivia", "Estonia"]
for country in countries:
    print(country)
````

````python
power = [2, 5, 36, 21]
energy = [kW/2 for kW in power]  # this is an example of python list comprehension. How would you right it in the conventional way as the cell above?
print(energy)
````

### While loops
While loops are control statements that allow part of a code to be executed repeatedly provided that a certain condition is being met.

````python
answer = input("What is the meaning of life?")  # this uses the input function which can be used to capture user input from the console
while answer != "42":
    print("Try again...")
    answer = input("What is the meaning of life?")
print("Correct")
````
